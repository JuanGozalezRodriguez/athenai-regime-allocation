# 03 · Perfilado "a contraluz" de los algoritmos (Bloque 2)

A partir solo de su serie: actividad, rentabilidad/riesgo, rendimiento por fase, estilo.

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent / "src"))

from athenai.config import load_config
from athenai.data.build import load_processed
from athenai.pipeline import run_profiling, run_regimes

cfg = load_config()
proc = load_processed(cfg)
reg = run_regimes(cfg, proc)
prof = run_profiling(cfg, proc, reg["dominant"])
```

## Distribución de estilos inferidos

```python
print(prof["styles"]["estilo"].value_counts())
```

## Fichas: los 10 algoritmos con mejor Sharpe

```python
cols = ["retorno_total", "vol_anual", "sharpe", "max_drawdown", "frac_en_mercado",
        "beta_mercado", "estilo"]
prof["fichas"].sort_values("sharpe", ascending=False)[cols].head(10).round(3)
```

## Rendimiento medio por estilo y fase de mercado

```python
by_regime = prof["by_regime"].join(prof["styles"]["estilo"])
by_regime.groupby("estilo").mean().round(4)
```
